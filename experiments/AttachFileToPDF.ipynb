# Store the structured md file inside the PDF as an attachment


## Question

>Can we not add a structured version of the input text version of a PDF (like a word/md/latex file) made eg via `pandoc` somehow to a PDF? In this way we keep the structural information for free, without needing to go to the compliecated tagged pdf route. 

## Answer: yes!

* Indeed, possible and easy with `pdftk`
* **Costs** The last cell here shows that adding a tiny 2.465 byte file to a pdf, enlarges the PDF with about 10%, in this case 50K. Crazy large!


## Result

A normal person does not see the difference, and will not see the attachment.

1. In preview I could not see the attached file back
2. In Chrome pdf viewer you can see the attacgment in the menu on the left and download it.
3. In Adobe you can see it using the "attachment" icon (like a paperclip), I could not save/export a txt.gz file, but this might be my settings.


## Workflow

1. Edit je document lekker
2. Ben je klaar? Dan
    1. Maak er een PDF van.
    2. Maak er een markdown of html of typst versie van met `pandoc` en gzip die.
    3. attach de gzipped structure versie aan de PDF met `pdftk`
3. Speciale screen readers zouden dan eerst naar die HTML/md/typst versie kunnen zoeken en als die er is, dan die laten zien i.p.v. de PDF.

### Mogelijke tegenwerping

* Ja, maardan geef ikd e source van mij PDF vrij, en kunnen mensen het zo weer veranderen.
* Tegenwerping daarop: ja joh, maar met `docling` kunnen ze ook aan die markdown of html komen. En PDFs zijn op zoveel manieren te veranderen.



## exiftool

Via exiftool kan je niet de atacgments zien, of eraan hangen, wat [Gemini ook zegt](https://www.google.com/search?q=exiftool+add+embedded+attachments)

```python
# see https://www.pdflabs.com/docs/pdftk-man-page/#dest-op-attach

# Attach a file to a pdf
!pdftk tagged_pdf.pdf attach_files tagged_pdf.txt.gz output tagged_pdf_Plus_attachment.pdf
```

```python
# get the attached file again, in directory effe

!mkdir effe
!pdftk tagged_pdf_Plus_attachment.pdf unpack_files output  effe
!ls -l effe
!rm -r effe
```

```python
!ls -l  tagged_pdf.pdf
!pdftotext tagged_pdf.pdf - |gzip > tagged_pdf.txt.gz
!ls -l tagged_pdf.txt.gz
!pdftk tagged_pdf.pdf attach_files tagged_pdf.txt.gz  output tagged_pdf_Plus_attachment.pdf
!ls -l tagged_pdf_Plus_attachment.pdf
```

```python
!exiftool -a -u -g1   tagged_pdf_Plus_attachment.pdf  
```

```python
!exiftool -Attach="tagged_pdf.txt.gz" -o tagged_pdf_Plus_attachmentViaExiftool.pdf tagged_pdf.pdf
```
